# 1. Dados de telemetria

São armazenados os dados simulados da telemetria da Aurora Siger. As informações representam diferentes situações que serão analisadas pelo sistema.

```python
dados_telemetria = {
    "1": {
        "data": "01/07/2026",
        "temperatura_interna": 36,
        "temperatura_externa": 18,
        "integridade_estrutural": 0,
        "energia": 65,
        "pressao_tanques": 85,
        "modulos_criticos": "FALHA"
    },

    "2": {
        "data": "01/08/2026",
        "temperatura_interna": 29,
        "temperatura_externa": 18,
        "integridade_estrutural": 1,
        "energia": 82,
        "pressao_tanques": 91,
        "modulos_criticos": "ALERTA"
    },

    "3": {
        "data": "01/09/2026",
        "temperatura_interna": 22,
        "temperatura_externa": 18,
        "integridade_estrutural": 1,
        "energia": 92,
        "pressao_tanques": 152,
        "modulos_criticos": "OK"
    }
}
```

# 2. Limites de segurança

São definidos os limites considerados seguros para cada parâmetro da telemetria. Esses valores serão utilizados pelo sistema para identificar possíveis falhas.

```python
# 🌡️ Temperatura interna
TEMP_INTERNA_MIN = 15
TEMP_INTERNA_MAX = 30

# 🌡️ Temperatura externa
TEMP_EXTERNA_MIN = -50
TEMP_EXTERNA_MAX = 50

# 🏗️ Integridade estrutural
INTEGRIDADE_OK = 1

# ⚡ Energia
ENERGIA_MIN = 80
ENERGIA_MAX = 100

# ⛽ Pressão dos tanques
PRESSAO_MIN = 150
PRESSAO_MAX = 300

# 🔧 Módulos críticos
MODULOS_OK = "OK"
```

# 3. Seleção da telemetria

permite selecionar qual registro de telemetria será analisado. O usuário escolhe uma das três datas disponíveis no sistema.

```python
def selecionar_telemetria():

    print()
    print("SELEÇÃO DA TELEMETRIA")

    print()
    print("1 - 01/09/2026")
    print("2 - 01/08/2026")
    print("3 - 01/07/2026")
    print()

    opcao = input("Digite uma opção: ").strip()

    if opcao not in dados_telemetria:
        print()
        print("Opção inválida.")
        return None

    return dados_telemetria[opcao]
```

# 4. Exibição dos dados

Os dados da telemetria selecionada são apresentados de forma organizada. Isso facilita a visualização dos parâmetros que serão avaliados.

```python
def exibir_telemetria(dados):

    print()
    print("=" * 60)
    print("DADOS DE TELEMETRIA")
    print("=" * 60)

    print(f" Data                     : {dados['data']}")
    print(f" Temperatura interna      : {dados['temperatura_interna']} °C")
    print(f" Temperatura externa      : {dados['temperatura_externa']} °C")
    print(f" Integridade estrutural   : {dados['integridade_estrutural']}")
    print(f" Nível de energia         : {dados['energia']} %")
    print(f" Pressão dos tanques      : {dados['pressao_tanques']} PSI")
    print(f" Módulos críticos         : {dados['modulos_criticos']}")
```

# 5. Diagnóstico operacional
cada parâmetro é comparado com seu respectivo limite de segurança. Quando um valor está fora do limite, o sistema registra o problema encontrado.

```python
def diagnosticar(dados):

    problemas = []

    print()
    print("=" * 60)
    print("DIAGNÓSTICO OPERACIONAL")
    print("=" * 60)

    # 🌡️ Temperatura interna

    if TEMP_INTERNA_MIN <= dados["temperatura_interna"] <= TEMP_INTERNA_MAX:
        print(" Temperatura interna: OK")
    else:
        print(" Temperatura interna: FALHA")
        problemas.append(
            f"Temperatura interna fora do limite "
            f"({TEMP_INTERNA_MIN}–{TEMP_INTERNA_MAX} °C): "
            f"{dados['temperatura_interna']} °C"
        )

    # 🌡️ Temperatura externa

    if TEMP_EXTERNA_MIN <= dados["temperatura_externa"] <= TEMP_EXTERNA_MAX:
        print(" Temperatura externa: OK")
    else:
        print(" Temperatura externa: FALHA")
        problemas.append(
            f"Temperatura externa fora do limite "
            f"({TEMP_EXTERNA_MIN}–{TEMP_EXTERNA_MAX} °C): "
            f"{dados['temperatura_externa']} °C"
        )

    # 🏗️ Integridade estrutural

    if dados["integridade_estrutural"] == INTEGRIDADE_OK:
        print(" Integridade estrutural: OK")
    else:
        print(" Integridade estrutural: FALHA")
        problemas.append(
            "Falha na integridade estrutural."
        )

    # ⚡ Energia

    if ENERGIA_MIN <= dados["energia"] <= ENERGIA_MAX:
        print(" Energia: OK")
    else:
        print(" Energia: FALHA")
        problemas.append(
            f"Nível de energia fora do limite "
            f"({ENERGIA_MIN}–{ENERGIA_MAX}%): "
            f"{dados['energia']}%"
        )

    # ⛽ Pressão dos tanques

    if PRESSAO_MIN <= dados["pressao_tanques"] <= PRESSAO_MAX:
        print(" Pressão dos tanques: OK")
    else:
        print(" Pressão dos tanques: FALHA")
        problemas.append(
            f"Pressão dos tanques fora do limite "
            f"({PRESSAO_MIN}–{PRESSAO_MAX} PSI): "
            f"{dados['pressao_tanques']} PSI"
        )

    # 🔧 Módulos críticos

    if dados["modulos_criticos"] == MODULOS_OK:
        print(" Módulos críticos: OK")
    else:
        print(" Módulos críticos: FALHA")
        problemas.append(
            f"Módulos críticos em estado "
            f"{dados['modulos_criticos']}."
        )

    return problemas
```

# 6. FUNÇÃO DO STATUS FINAL
Reúne os resultados do diagnóstico e determina o estado final da simulação. Se nenhum problema for encontrado, o sistema indica “PRONTO PARA DECOLAR”; caso contrário, indica “DECOLAGEM ABORTADA”.

```python
def exibir_status_final(problemas):

    print()
    print("=" * 60)
    print("STATUS FINAL")
    print("=" * 60)

    if len(problemas) == 0:

        print()
        print("Nenhum problema detectado.")
        print()
        print(">>> PRONTO PARA DECOLAR <<<")

    else:

        print()
        print("PROBLEMAS DETECTADOS:")

        for numero, problema in enumerate(problemas, start=1):
            print(f" {numero} - {problema}")

        print()
        print(f"TOTAL DE PROBLEMAS: {len(problemas)}")

        print()
        print(">>> DECOLAGEM ABORTADA <<<")
```

# EXECUÇÃO COMPLETA
Etapa final do programa. Ela chama todas as funções na ordem correta, permitindo executar o sistema completo de uma única vez.

```python
print()
print("=" * 60)
print("INICIANDO SISTEMA AURORA SIGER")
print("=" * 60)

while True:

    # 3.SELEÇÃO DA TELEMETRIA
    dados = selecionar_telemetria()
    if dados is None:
        continue

    # 4. EXIBIÇÃO DOS DADOS
    exibir_telemetria(dados)

    # 5.DIAGNÓSTICO
    problemas = diagnosticar(dados)

    # 6. STATUS FINAL
    exibir_status_final(problemas)

    # PERGUNTAR NOVAMENTE
    print()
    print("=" * 60)
    print("NOVA CONSULTA")
    print("=" * 60)

    novamente = input(
        "Deseja analisar outra telemetria? (S/N): "
    ).strip().upper()

    if novamente != "S":
        break

# ENCERRAMENTO

print()
print("=" * 60)
print("SISTEMA ENCERRADO")
print("EXECUÇÃO CONCLUÍDA")
print("=" * 60)
```
