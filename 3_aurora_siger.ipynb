# 🚀 AURORA SIGER

## 1. Inicialização
apresentação inicial do programa, criando uma função que mostra no terminal que o sistema de telemetria está começando.

```python
def inicialização():
    print("==================================================")
    print(" 1.INICIALIZAÇÃO ")
    print("==================================================")
    print(" ")
    print("INICIAR sistema de telemetria")
    print(" ")
```

## 2. Telemetria
O programa gera automaticamente valores aleatórios para representar os sensores e, em seguida, apresenta essas informações de forma organizada no terminal.

> import random — Importa a biblioteca random, que permite gerar valores aleatórios para simular os dados dos sensores.


```python
import random
from datetime import datetime

def dados_telemetria():
    return {
        "DATA": datetime.now().strftime("%d/%m/%Y"),
        "T_INT": round(random.uniform(10, 35), 2),
        "T_EXT": round(random.uniform(-60, 60), 2),
        "STR": random.randint(0, 1),
        "PWR": round(random.uniform(50, 100), 2),
        "PRS": round(random.uniform(100, 350), 2),
        "MOD": random.choice(["OK", "FALHA"])
    }

def exibir_telemetria(dados):
    print("==================================================")
    print(" 2.RECEBIMENTO DOS DADOS")
    print("==================================================")
    print(" ")
    print(f"Data:{dados['DATA']}")
    print(f"Temperatura interna: {dados['T_INT']}")
    print(f"Temperatura externa: {dados['T_EXT']}")
    print(f"Integridade estrutural: {dados['STR']}")
    print(f"Nível de energia: {dados['PWR']}")
    print(f"Pressão dos tanques: {dados['PRS']}")
    print(f"Módulos críticos: {dados['MOD']}")
```

## 3. Validação dos Parâmetros

Cada parâmetro é comparado com os seus limites aceitáveis. Quando um valor está correto, aparece OK; quando está fora dos limites, aparece NOK e a falha é registrada.


```python
def validar_parametros(dados):

    print()
    print("==================================================")
    print("3. VALIDAÇÃO DOS PARÂMETROS")
    print("==================================================")
    print(" ")

    falhas = []
    sistema_ok = True

    # Temperatura interna
    if 15 <= dados["T_INT"] <= 30:
        print("Temperatura interna: OK")
    else:
        print("Temperatura interna: NOK")
        falhas.append("Temperatura interna")
        sistema_ok = False

    # Temperatura externa
    if -50 <= dados["T_EXT"] <= 50:
        print("Temperatura externa: OK")
    else:
        print("Temperatura externa: NOK")
        falhas.append("Temperatura externa")
        sistema_ok = False

    # Integridade estrutural
    if dados["STR"] == 1:
        print("Integridade estrutural: OK")
    else:
        print("Integridade estrutural: NOK")
        falhas.append("Integridade estrutural")
        sistema_ok = False

    # Nível de energia
    if 80 <= dados["PWR"] <= 100:
        print("Nível de energia: OK")
    else:
        print("Nível de energia: NOK")
        falhas.append("Nível de energia")
        sistema_ok = False

    # Pressão dos tanques
    if 150 <= dados["PRS"] <= 300:
        print("Pressão dos tanques: OK")
    else:
        print("Pressão dos tanques: NOK")
        falhas.append("Pressão dos tanques")
        sistema_ok = False

    # Módulos críticos
    if dados["MOD"] == "OK":
        print("Módulos críticos: OK")
    else:
        print("Módulos críticos: NOK")
        falhas.append("Módulos críticos")
        sistema_ok = False

    return sistema_ok, falhas
```

## 4. Decisão Final

o programa toma a decisão final com base na validação dos parâmetros. Se todos os parâmetros estiverem corretos, o lançamento é autorizado. Caso exista pelo menos uma falha, o programa mostra a quantidade e quais foram as falhas encontradas e informa que o lançamento não está autorizado.
> for falha in falhas: — Percorre todos os elementos existentes na lista falhas, analisando cada falha individualmente.

```python
def decisao_final(sistema_ok, falhas):

    print()
    print("==================================================")
    print("4. DECISÃO FINAL")
    print("==================================================")
    print(" ")

    if sistema_ok == True:

        print("STATUS: DECOLAR")

    else:

        print(f"Número de falhas: {len(falhas)}")
        print()
        print("Falhas identificadas:")
        for falha in falhas:
            print(f"- {falha}")
        print()
        print("STATUS: ABORTAR")
```

# 5. Programa

Esta célula funciona como o fluxo principal do programa, executando todas as etapas na ordem correta. Primeiro o sistema é inicializado, depois os dados são recebidos, em seguida são validados e, por fim, é tomada a decisão final sobre a autorização do lançamento.

```python
# 1.INICIALIZAÇÃO
inicialização()

# 2.RECEBIMENTO DOS DADO
dados = dados_telemetria()
exibir_telemetria(dados)

# 3.VALIDAÇÃO DOS PARÂMETROS
sistema_ok, falhas = validar_parametros(dados)

# 4. DECISÃO FINAL
decisao_final(sistema_ok, falhas)
```
