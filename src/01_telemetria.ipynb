```python

# ============================================================
#              AURORA SIGER - SISTEMA DE TELEMETRIA
# ============================================================


# ============================================================
#                         DADOS
# ============================================================

dados_telemetria = {
    "1": {
        "data": "01/09/2026",
        "temperatura_interna": 22,
        "temperatura_externa": 18,
        "integridade_estrutural": 1,
        "energia": 92,
        "pressao_tanques": 152,
        "modulos_criticos": "OK"
    },

    "2": {
        "data": "01/08/2026",
        "temperatura_interna": 29,
        "temperatura_externa": 18,
        "integridade_estrutural": 1,
        "energia": 82,
        "pressao_tanques": 91,
        "modulos_criticos": "ALERTA"
    },

    "3": {
        "data": "01/07/2026",
        "temperatura_interna": 36,
        "temperatura_externa": 18,
        "integridade_estrutural": 0,
        "energia": 65,
        "pressao_tanques": 85,
        "modulos_criticos": "FALHA"
    }
}


# ============================================================
#                         MENU
# ============================================================

while True:

    print()
    print("=" * 60)
    print("AURORA SIGER - TELEMETRIA")
    print("=" * 60)

    print()
    print("Selecione a data da telemetria:")
    print()

    print("1 - 01/09/2026")
    print("2 - 01/08/2026")
    print("3 - 01/07/2026")
    print("4 - Sair")

    print()

    opcao = input("Digite uma opção: ").strip()


    # ========================================================
    #                         SAIR
    # ========================================================

    if opcao == "4":

        print()
        print("=" * 60)
        print("Sistema encerrado.")
        print("=" * 60)

        break


    # ========================================================
    #                  VERIFICAR OPÇÃO
    # ========================================================

    if opcao not in dados_telemetria:

        print()
        print("Opção inválida. Digite 1, 2, 3 ou 4.")
        continue


    dados = dados_telemetria[opcao]


    # ========================================================
    #                 DADOS DE TELEMETRIA
    # ========================================================

    print()
    print("=" * 60)
    print("DADOS DE TELEMETRIA")
    print("=" * 60)

    print(f" Data                     : {dados['data']}")
    print(f" Temperatura interna      : {dados['temperatura_interna']} °C")
    print(f" Temperatura externa      : {dados['temperatura_externa']} °C")
    print(f" Integridade estrutural   : {dados['integridade_estrutural']}")
    print(f" Nível de energia         : {dados['energia']} %")
    print(f" Pressão dos tanques      : {dados['pressao_tanques']} PSI")
    print(f" Módulos críticos         : {dados['modulos_criticos']}")


    # ========================================================
    #              DIAGNÓSTICO DA ESTAÇÃO
    # ========================================================

    problemas = []

    print()
    print("=" * 60)
    print("DIAGNÓSTICO DA ESTAÇÃO")
    print("=" * 60)


    # --------------------------------------------------------
    # TEMPERATURA INTERNA
    # Mínimo: 15 °C
    # Máximo: 30 °C
    # --------------------------------------------------------

    if 15 <= dados["temperatura_interna"] <= 30:

        print(" Temperatura interna: OK")

    else:

        print(" Temperatura interna: FALHA")

        problemas.append(
            f"Temperatura interna fora do limite "
            f"(15–30 °C): {dados['temperatura_interna']} °C"
        )


    # --------------------------------------------------------
    # TEMPERATURA EXTERNA
    # Mínimo: -50 °C
    # Máximo: 50 °C
    # --------------------------------------------------------

    if -50 <= dados["temperatura_externa"] <= 50:

        print(" Temperatura externa: OK")

    else:

        print(" Temperatura externa: FALHA")

        problemas.append(
            f"Temperatura externa fora do limite "
            f"(-50–50 °C): {dados['temperatura_externa']} °C"
        )


    # --------------------------------------------------------
    # INTEGRIDADE ESTRUTURAL
    # Deve ser: 1
    # --------------------------------------------------------

    if dados["integridade_estrutural"] == 1:

        print(" Integridade estrutural: OK")

    else:

        print(" Integridade estrutural: FALHA")

        problemas.append(
            "Falha na integridade estrutural."
        )


    # --------------------------------------------------------
    # ENERGIA
    # Mínimo: 80%
    # Máximo: 100%
    # --------------------------------------------------------

    if 80 <= dados["energia"] <= 100:

        print(" Energia: OK")

    else:

        print(" Energia: FALHA")

        problemas.append(
            f"Nível de energia fora do limite "
            f"(80–100%): {dados['energia']}%"
        )


    # --------------------------------------------------------
    # PRESSÃO DOS TANQUES
    # Mínimo: 150 PSI
    # Máximo: 300 PSI
    # --------------------------------------------------------

    if 150 <= dados["pressao_tanques"] <= 300:

        print(" Pressão dos tanques: OK")

    else:

        print(" Pressão dos tanques: FALHA")

        problemas.append(
            f"Pressão dos tanques fora do limite "
            f"(150–300 PSI): {dados['pressao_tanques']} PSI"
        )


    # --------------------------------------------------------
    # MÓDULOS CRÍTICOS
    # Deve ser: OK
    # --------------------------------------------------------

    if dados["modulos_criticos"] == "OK":

        print(" Módulos críticos: OK")

    else:

        print(" Módulos críticos: FALHA")

        problemas.append(
            f"Módulos críticos em estado "
            f"{dados['modulos_criticos']}."
        )


    # ========================================================
    #                    STATUS FINAL
    # ========================================================

    print()
    print("=" * 60)
    print("STATUS FINAL")
    print("=" * 60)

    quantidade_problemas = len(problemas)


    if quantidade_problemas == 0:

        print()
        print("Nenhum problema detectado.")

    else:

        print()
        print("PROBLEMAS DETECTADOS:")

        for numero, problema in enumerate(problemas, start=1):

            print(f" {numero} - {problema}")


    print()
    print(f"PROBLEMAS DETECTADOS: {quantidade_problemas}")

```
